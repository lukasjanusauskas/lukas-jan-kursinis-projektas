```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

plt.style.use('bmh')
plt.rcParams['axes.facecolor'] = 'white'
```

```python
y = np.load('../anom-y.npy')
y_low = np.load('../anom-y_pred_low.npy', allow_pickle=True)
y_high = np.load('../anom-y_pred_high.npy', allow_pickle=True)
```

### COG pokyčio atsakas

```python
plt.plot(y[:, 0, 0])
plt.plot(y_low[:, 0, 0])
plt.plot(y_high[:, 0, 0])
```

```python
plt.plot(y[:450, 0, 0])
plt.plot(y_low[:450, 0, 0])
plt.plot(y_high[:450, 0, 0])
```

```python
plt.plot(y[450:550, 0, 0])
plt.plot(y_low[450:550, 0, 0])
plt.plot(y_high[450:550, 0, 0])
```

```python
plt.plot(y[550:, 0, 0])
plt.plot(y_low[550:, 0, 0])
plt.plot(y_high[550:, 0, 0])
```

```python
count_mistakes = np.sum( 
    (y[450:, 0, 0] < y_low[450:, 0, 0]) |
    (y[450:, 0, 0] > y_high[450:, 0, 0])
)

count_total = y[450:, 0, 0].shape[0]

1 - (count_mistakes / count_total)
```
