```python
X_arr = np.load('../anom-x.npy', allow_pickle=True)
y_arr = np.load('../anom-y.npy', allow_pickle=True)

n_time = X_arr.shape[1]
n_inp = X_arr.shape[2]
n_out = y_arr.shape[2]

X_arr = X_arr.reshape((-1, n_inp*n_time))
```

```python
import pickle
with open('../output/qrf-final.pkl', 'rb+') as f:
    qrf_model = pickle.load(f)
```

```python
y_out = qrf_model.predict(X_arr, quantiles=[0.025, 0.975])
```

```python
from datetime import datetime
timestamps = list(pd.date_range(start=datetime(2024,11,18,0,0), end=datetime(2024,11,19,0,0), freq='60s')[31:-26])
```

```python
y_flat = y_arr[:, 0, 0]
y_low = y_out[:, :, 0]\
    .reshape((-1, 25, 2))\
    [:, 0, 0]
y_high = y_out[:, :, 1]\
    .reshape((-1, 25, 2))\
    [:, 0, 0]

anomalies_x = []
anomalies_y = []
for i in range(y_flat.shape[0]):

    if y_flat[i] < y_low[i]:
        anomalies_x.append( timestamps[i] )
        anomalies_y.append( y_flat[i] )

    elif y_flat[i] > y_high[i]:
        anomalies_x.append( timestamps[i] )
        anomalies_y.append( y_flat[i] )

plt.plot(timestamps, y_flat, label = "Target value", color='black', alpha=0.6)
plt.plot(timestamps, y_low, color='#185288', linestyle='--', linewidth=2)
plt.plot(timestamps, y_high, color="#185288", linestyle='--')

plt.xticks(
    ticks=[datetime(2024,11,18, 3+x*6,0) for x in range(0, 4)],
    labels=[f"2024-11-18\n{(3+x*6):02}:00" for x in range(0, 4)],
)
plt.tick_params(axis='both', which='both', labelsize=13)

plt.xlabel('Timestamp (UTC)', size=18)
plt.ylabel(r'$\Delta COG$', size=18)
plt.title(r'Yi-Peng 3 $\Delta COG$ PI one minute ahead', size=20)

plt.fill_between(timestamps, y1=y_low, y2=y_high, alpha=0.4, label='Prediction interval')
plt.scatter(anomalies_x, anomalies_y, color='red', s=20, zorder=5, label='Anomalies')
plt.legend(fontsize=13)

plt.tight_layout()
plt.savefig("cog-difference-res-qrf.png", dpi=750)

plt.show()
```

```python
df_anom = pd.read_csv(
    '../data/anomaly_meteo.csv', 
    index_col=['Unnamed: 0'],
    parse_dates=['position_timestamp']
)
df_anom.info()

df_anom.drop(columns=['windWaveDirection', 'windWaveHeight'], inplace=True)

df_anom = df_anom.rename(columns={
    'course': 'COG',
    'heading': 'Heading',
    'position_timestamp': '# Timestamp',
    'latitude': 'Latitude',
    'longitude': 'Longitude',
    'speed': 'SOG'
})
df_anom['MMSI'] = 414270000

dfs = []
above_interpolation_limit = 0

df_anom['Heading'] = np.clip( df_anom['Heading'], 0, 360 )
df_anom['COG'] = np.clip( df_anom['COG'], 0, 360 )

df_anom['hour'] = df_anom['# Timestamp'].dt.round('h')
df_anom = df_anom.drop(columns=['hour'])

df_anom, timestamps = feature_engineering(df_anom, return_timestamps=True)

min_max_scale_cols = [
    'Latitude',
    'Longitude',
    'SOG',
    'currentSpeed',
    'gust',
    'swellHeight',
    'waveHeight',
    'windSpeed'
]

with open('../output/lstm-min-max-scaler.pkl', 'rb') as f:
    scaler = pickle.load(f)

df_anom[min_max_scale_cols] = scaler.fit_transform(
    df_anom[min_max_scale_cols].values
)

df_anom.dropna(inplace=True)
df_anom.drop(columns=['time'], inplace=True)

X_arr, y_arr = prepare_time_series(df_anom)
```
