```python
import pickle

with open('../output/history_low-final.pkl', 'rb+') as f:
    history_low = pickle.load(f)
with open('../output/history_high-final.pkl', 'rb+') as f:
    history_high = pickle.load(f)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.style.use('bmh')
plt.rcParams['axes.facecolor'] = 'white'
```

```python
plt.plot(
    history_low['loss'],
    label=r'$\rho_t$ mokymo aibėje'
)
plt.plot(
    history_low['val_loss'],
    label=r'$\rho_t$ validavimo aibėje'
)

plt.ylabel('Kvantilių nuostolių\nfunkcijos reikšmės', size=18)
plt.xlabel('Epocha', size=18)
plt.tick_params(axis='both', labelsize=14)
plt.title('Kvantilių nuostolių funkcijos reikšmės\nLSTM-AE apatinės ribos modeliui', size=20)

plt.legend(fontsize=16)

plt.tight_layout()
plt.savefig('epochos-low.png', dpi=750)
```

```python
plt.plot(
    history_high['loss'],
    label=r'$\rho_t$ mokymo aibėje'
)
plt.plot(
    history_high['val_loss'],
    label=r'$\rho_t$ validavimo aibėje'
)

plt.ylabel('Kvantilių nuostolių\nfunkcijos reikšmės', size=18)
plt.xlabel('Epocha', size=18)
plt.tick_params(axis='both', labelsize=14)
plt.title('Kvantilių nuostolių funkcijos reikšmės\nLSTM-AE viršutinės ribos modeliui', size=20)

plt.legend(fontsize=16)

plt.tight_layout()
plt.savefig('epochos-high.png', dpi=750)
```
